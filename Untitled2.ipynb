```python
#what is a neuron
# a neuron is unit which take input(s) and deliver outputs and with each input we attach weights and add a bias term to the whole value
```

```python
input = [6, 3, 2]
weights = [0.5, -0.5, -0.8]
bias = 0.4
output = input[0]*weights[0] + input[1]*weights[1] + input[2]*weights[2] + bias
print(output)
```
